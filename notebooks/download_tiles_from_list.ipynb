# Download tiles from list

```python
import os
import requests
import csv
from tqdm import tqdm

SRC_TILES = r"D:\GitHubProjects\Terranum_repo\facet_orientation\brienz\list_tiles.csv"
SRC_DEST_DIR = os.path.join(os.path.dirname(SRC_TILES), "data")

os.makedirs(SRC_DEST_DIR, exist_ok=True)
def download_tile(url_tile, src_dest):
    data = requests.get(url_tile).content
    with open(src_dest, 'wb') as handler:
        handler.write(data)


with open(SRC_TILES, 'r') as f:
    reader = csv.reader(f)
    data = [x[0] for x in list(reader)]

for _, link in tqdm(enumerate(data), total=len(data)):
    src_dest = os.path.join(SRC_DEST_DIR, link.split('/')[-1])
    download_tile(link, src_dest)
```

# unzip all files in folder

```python
import os
import zipfile

SRC_FOLDER = r"D:\GitHubProjects\Terranum_repo\facet_orientation\brienz\data"

list_of_files = [os.path.join(SRC_FOLDER, x) for x in os.listdir(SRC_FOLDER) if x.endswith('.zip')]

for _, src_file in tqdm(enumerate(list_of_files), total=len(list_of_files)):
    with zipfile.ZipFile(src_file, 'r') as zip_ref:
        zip_ref.extractall(SRC_FOLDER)
    os.remove(src_file)
```
